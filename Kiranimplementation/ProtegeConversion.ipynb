```python
import csv
from rdflib import Graph, Literal, Namespace, URIRef
from rdflib.namespace import RDF, RDFS

# Define namespaces
MITRE = Namespace("http://kngbq.org/CKG/#")

# Load CSV data
csv_file = './attack_groups.csv'  # Replace with your CSV file path
rdf_graph = Graph()

# Read CSV and convert to RDF
with open(csv_file, 'r') as csvfile:
    csv_reader = csv.DictReader(csvfile)
    for row in csv_reader:
        # Create URI for the resource
        resource_uri = URIRef(f"http://kngbq.org/CKG/attack_groups/{row['ID']}")  # You can customize the URI as needed

        # Add triples to the RDF graph
        rdf_graph.add((resource_uri, RDF.type, MITRE.AttackGrp))  # Assuming Technique is a class in your ontology
        rdf_graph.add((resource_uri, RDFS.label, Literal(row['Name'])))  # Assuming 'Label' is a column in your CSV

# Serialize RDF graph to RDF/XML format
rdf_output = rdf_graph.serialize(format='xml')

# Save RDF data to a file
with open('attack_groups.rdf', 'wb') as output_file:
    output_file.write(rdf_output.encode('utf-8'))
```

# Convert CSV to TSV

```python
import csv

def csv_to_tsv(input_csv, output_tsv):
    with open(input_csv, 'r') as csv_file:
        csv_reader = csv.reader(csv_file)
        rows = list(csv_reader)

    with open(output_tsv, 'w', newline='') as tsv_file:
        tsv_writer = csv.writer(tsv_file, delimiter='\t')
        tsv_writer.writerows(rows)

# Specify your CSV and TSV file paths
input_csv_file = 'attack_groups.csv'
output_tsv_file = 'attack_groups.tsv'

# Convert CSV to TSV
csv_to_tsv(input_csv_file, output_tsv_file)
```
