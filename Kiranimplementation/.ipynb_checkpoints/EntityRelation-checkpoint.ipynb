```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Spoofing: Unauthorized access to the power grid system by impersonating a legitimate user or device.Tampering: Unauthorized modification of system components or data, leading to potential malfunctions or disruptions.Repudiation: Users or entities denying their actions within the system, making it difficult to trace malicious activities.Information Disclosure: Unauthorized access to sensitive information within the power grid system, leading to potential exploitation.")

for ent in doc.ents:
    print(ent.text, ent.label_)
```
