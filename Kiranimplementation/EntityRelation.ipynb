```python
import spacy
from spacy.training.example import Example
from spacy.training.iob_utils import offsets_to_biluo_tags

# Load a blank English model
nlp = spacy.blank("en")

# Add a named entity label for the threat type
ner = nlp.add_pipe("ner")
ner.add_label("THREAT_TYPE")

# Training data with examples
training_data = [
    ("Spoofing: Unauthorized access to the power grid system by impersonating a legitimate user or device.", {"entities": [(0, 8, "THREAT_TYPE")]}),
    ("Tampering: Unauthorized modification of system components or data, leading to potential malfunctions or disruptions.", {"entities": [(0, 8, "THREAT_TYPE")]}),
    ("Repudiation: Users or entities denying their actions within the system, making it difficult to trace malicious activities.", {"entities": [(0, 11, "THREAT_TYPE")]}),
    ("Information Disclosure: Unauthorized access to sensitive information within the power grid system, leading to potential exploitation.", {"entities": [(0, 20, "THREAT_TYPE")]})
]

for text, entities in training_data:
    doc = nlp.make_doc(text)
    biluo_tags = offsets_to_biluo_tags(doc, entities["entities"])
    print(f"Text: {text}")
    print(f"BILUO Tags: {biluo_tags}")
    print()
```
